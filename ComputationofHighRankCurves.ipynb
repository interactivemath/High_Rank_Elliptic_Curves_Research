```python
from sympy import Matrix
def f(x,y):
    return [x*y,y,-x^2,-x,-1,x*y^2,x^2*y,y^2,-x^3,0] #the intuition behind this was using a traditional cubic and then moving all the terms to one size to have it equal to 0
M=Matrix.zeros(7,10) #7x10 matrix
for i in range(0,7):
    x = int(input("Enter a rational number:"))
    y = int(input("Enter a rational number:"))
    M[i,:] = [f(x,y)] #sets the row of the matrix to the function output
print(M.rref()) #solves the matrix --> reduced row echelon form
```
